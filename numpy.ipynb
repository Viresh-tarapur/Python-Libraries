```python
import numpy as np
print("NumPy Important Functions")
```

```python
# Array creation
a = np.array([1, 2, 3, 4])
print(a)
```

```python
# arange
print(np.arange(1, 10, 2))
```

```python
# linspace
print(np.linspace(0, 1, 5))
```

```python
# zeros and ones
print(np.zeros((2, 3)))
print(np.ones((2, 3)))
```

```python
# reshape
b = np.array([1, 2, 3, 4, 5, 6])
print(b.reshape(2, 3))
```

```python
# ndim, shape, size
c = np.array([[1, 2], [3, 4]])
print(c.ndim)
print(c.shape)
print(c.size)
```

```python
# Indexing and slicing
d = np.array([10, 20, 30, 40])
print(d[1])
print(d[1:3])
```

```python
# Mathematical operations
x = np.array([1, 2, 3])
y = np.array([4, 5, 6])
print(x + y)
print(x * y)
print(x ** 2)
```

```python
# sum, mean, max, min
e = np.array([[1, 2, 3], [4, 5, 6]])
print(np.sum(e))
print(np.mean(e))
print(np.max(e))
print(np.min(e))
```

```python
# axis usage
print(np.sum(e, axis=0))
print(np.sum(e, axis=1))
```

```python
# where
f = np.array([10, 20, 30, 40])
print(np.where(f > 25, f, 0))
```

```python
# sort and argsort
g = np.array([40, 10, 30, 20])
print(np.sort(g))
print(np.argsort(g))
```

```python
# unique
h = np.array([1, 2, 2, 3, 3, 3])
print(np.unique(h))
```

```python
# concatenate
i = np.array([1, 2])
j = np.array([3, 4])
print(np.concatenate((i, j)))
```

```python
# transpose
k = np.array([[1, 2, 3], [4, 5, 6]])
print(k.T)
```

```python
# dot product
m = np.array([[1, 2], [3, 4]])
n = np.array([[5, 6], [7, 8]])
print(np.dot(m, n))
```

```python
# astype
p = np.array([1.2, 2.5, 3.8])
print(p.astype(int))
```

```python
# random functions
print(np.random.rand(3))
print(np.random.randint(1, 10, 5))
```
