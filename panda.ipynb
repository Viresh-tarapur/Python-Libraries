```python
import pandas as pd

# Creating DataFrame

data = {
    "Name": ['Ram', 'Shyam', 'Ghanshyam', 'Dhanshyam', 'Aditi', 'Jagdish', 'Raj', 'Simran'],
    "Age": [28, 34, 22, 30, 29, 40, 25, 32],
    "Salary": [50000, 60000, 45000, 52000, 49000, 70000, 48000, 58000],
    "Performance Score": [85, 90, 78, 92, 88, 95, 80, 89]
}

my_file = pd.DataFrame(data)

```

```python
# Saving to CSV

my_file.to_csv("practices.csv", index=False)

# Reading CSV

new_file = pd.read_csv("practices.csv")
print(" Data read from CSV file:")
print(new_file)

```

```python

# head() → First 5 rows

print(" First 5 rows (head):")
print(my_file.head())



```

```python

# tail() → Last 5 rows

print(" Last 5 rows (tail):")
print(my_file.tail())


```

```python

# info() → Structure of DataFrame

print(" DataFrame Info:")
my_file.info()

```

```python

# describe() → Statistical summary

print(" Statistical Description:")
print(my_file.describe())

```

```python


# shape → Rows & Columns

print(" Shape (Rows, Columns):")
print(my_file.shape)

```

```python

# columns → Column names

print(" Column Names:")
print(my_file.columns)


```

```python

# Single column selection
print(" Age Column:")
age_column = my_file["Age"]
print(age_column)


```

```python
 #Multiple column selection

print(" Age and Salary Columns:")
multi_columns = my_file[["Age", "Salary"]]
print(multi_columns)

```

```python

# Filtering rows (Age > 30)

print(" Employees with Age > 30:")
condition = my_file[my_file["Age"] > 30]
print(condition)

```

```python

# Filtering rows (Age > 30 AND Salary > 8000)

print(" Age > 30 AND Salary > 8000:")
condition1 = my_file[(my_file["Age"] > 30) & (my_file["Salary"] > 8000)]
print(condition1)

```

```python

print(" Age > 30 OR Performance Score > 90:")
filter_rows_2 = my_file[(my_file["Age"] > 30) | (my_file["Performance Score"] > 90)]
print(filter_rows_2)
```
